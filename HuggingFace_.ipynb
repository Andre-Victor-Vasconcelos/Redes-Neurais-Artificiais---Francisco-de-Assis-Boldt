```python
import random
from transformers import pipeline
from datasets import load_dataset
```

```python
# ==========================================
# 1. DOWNLOAD DA BASE IMDB (VIA HUGGING FACE)
# ==========================================
print("Baixando o dataset IMDB via Hugging Face Datasets...")

dataset = load_dataset("stanfordnlp/imdb", split="test")
```

```python
# ==========================================
# 2. SEPARANDO E CONVERTENDO EM LISTA DO PYTHON
# ==========================================
amostra = dataset.shuffle(seed=42).select(range(50))

textos = list(amostra['text'])
labels_reais = ["POSITIVE" if label == 1 else "NEGATIVE" for label in amostra['label']]

print("\nBase de dados carregada e convertida com sucesso!")
```

```python
# ==========================================
# 3. HUGGING FACE PIPELINE (ANALISADOR DE SENTIMENTO)
# ==========================================
print("Baixando o modelo de IA (DistilBERT)...")
analisador = pipeline("sentiment-analysis", truncation=True, max_length=512)
```

```python
# ==========================================
# 4. INFERÊNCIA E AVALIAÇÃO
# ==========================================
print("\nAnalisando os sentimentos dos textos...\n")
previsoes = analisador(textos)

# Calculando a acurácia
acertos = 0
for i in range(len(textos)):
    sentimento_previsto = previsoes[i]['label']
    confianca = previsoes[i]['score']
    sentimento_real = labels_reais[i]

    if sentimento_previsto == sentimento_real:
        acertos += 1

    if i < 3 or i >= 47:
        print(f"[{sentimento_real}] Texto: {textos[i][:60]}... -> IA previu: {sentimento_previsto} (Conf. {confianca:.2f})")

    if i == 3: print("...\n[Ocultando textos intermediários para não poluir o terminal]\n...")

acuracia = (acertos / len(textos)) * 100
print(f"\n======================================")
print(f"ACURÁCIA DA IA NA AMOSTRA: {acuracia:.1f}%")
print(f"======================================")
```
